```python
# ==========================================
# NUMPY: THE ABSOLUTE BASICS (To'liq qo'llanma)
# ==========================================

import numpy as np

# ------------------------------------------
# 1. Array yaratish va shape (shakl)
# ------------------------------------------
arr = np.array([1, 3, 5, 7])
print('numpy_array:', arr)
print(arr.shape)  # Natija: (4,)

arr2 = np.array([[1, 3, 5, 7], [4, 6, 8, 9]])
print('numpy_array2:', arr2)
print(arr2.shape)  # Natija: (2, 4)

# ------------------------------------------
# 2. Ma'lumot turi (dtype) va astype()
# ------------------------------------------
a = np.array([1, 2, 3])
print("dtype:", a.dtype)  # Natija: int64

a = np.array([1, 2, 3])
b = a.astype(float)
print("float:", b)  # Natija: [1. 2. 3.]

# ------------------------------------------
# 3. Shape'ni o'zgartirish: reshape()
# ------------------------------------------
a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2, 3)
print("shaped_array:\n", b)

# ------------------------------------------
# 4. Nol va birlar bilan to'ldirish: zeros(), ones()
# ------------------------------------------
z = np.zeros((2, 3))
o = np.ones((2, 3))

print("zeros:\n", z)
print("ones:\n", o)

# ------------------------------------------
# 5. O'lchamlar soni: ndim
# ------------------------------------------
a = np.array([[1, 2], [3, 4]])
print("ndim:", a.ndim)  # Natija: 2

# ------------------------------------------
# 6. Ketma-ketlik yaratish: arange()
# ------------------------------------------
a = np.arange(1, 7)
print("arange:", a)  # Natija: [1 2 3 4 5 6]

# ------------------------------------------
# 7. Statistik funksiyalar: sum(), mean(), max(), min()
# ------------------------------------------
a = np.array([1, 2, 3, 4])

print("sum:", a.sum())    # Natija: 10
print("mean:", a.mean())  # Natija: 2.5
print("max:", a.max())    # Natija: 4
print("min:", a.min())    # Natija: 1

# ------------------------------------------
# 8. Ko'p o'lchamli arrayni 1D ga o'tkazish: flatten()
# ------------------------------------------
a = np.array([[1, 2], [3, 4]])
print("flatten:", a.flatten())  # Natija: [1 2 3 4]

# ------------------------------------------
# 9. Transponirlash (Ustun va qator o'rnini almashtirish): .T
# ------------------------------------------
a = np.array([[1, 2], [3, 4]])
print("transpose:\n", a.T)

# ------------------------------------------
# 10. Matritsali ko'paytirish: dot() / @
# ------------------------------------------
a = np.array([[1, 2], [3, 4]])
b = np.array([[2, 0], [1, 2]])
print("dot/multiplication:\n", a.dot(b))

# ------------------------------------------
# 11. Tasodifiy sonlar yaratish: random.rand()
# ------------------------------------------
# 0.0 va 1.0 oralig'ida (0.0 kiradi, 1.0 kirmaydi)
a = np.random.rand(2, 3)
print("random rand:\n", a)
```
