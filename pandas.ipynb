```python
# ==========================================
# PANDAS: TO'LIQ QO'LLANMA (Titanic dataset va asoslar)
# ==========================================

import pandas as pd

# Pandas versiyasini tekshirish
print("Pandas versiyasi:", pd.__version__)

# ------------------------------------------
# 2. DataFrame yaratish va asosiy amallar
# ------------------------------------------
mydataset = {
    'cars': ["BMW", "Volvo", "Ford", "Tesla", "Ferrari", "Ford", "BMW"],
    'counts': [3, 7, 2, 8, 1, 9, 3]
}

df = pd.DataFrame(mydataset)
print("\nYaratilgan DataFrame:")
print(df)

# Boshlang'ich va oxirgi qatorlar
print("\n--- df.head() ---")
print(df.head())

print("\n--- df.tail() ---")
print(df.tail())

# Statistik ma'lumotlar
print("\n--- df.describe() ---")
print(df.describe())

print("\n--- df.count() ---")
print(df.count())

print("\n--- df.std(numeric_only=True) ---")
print(df.std(numeric_only=True))

# ------------------------------------------
# 3. Indeks va ustunlar bilan ishlash (iloc va loc)
# ------------------------------------------
print("\n--- df.iloc[2:] ---")
print(df.iloc[2:])

print("\n--- df.loc[2:5, 'cars'] ---")
print(df.loc[2:5, "cars"])

# ------------------------------------------
# 4. Ma'lumotlarni filtrlash
# ------------------------------------------
print("\n--- df[df['counts'] > 3] ---")
print(df[df["counts"] > 3])

print("\n--- df[df['cars'] == 'Tesla'] ---")
print(df[df["cars"] == "Tesla"])

# ------------------------------------------
# 5. Yo'qotilgan (Missing / NaN) ma'lumotlar bilan ishlash
# ------------------------------------------
data_with_missing = {
    "cars": ["BMW", "Volvo", None, "Tesla", "Ferrari", None, "Ford"],
    "counts": [3, 7, None, 8, 1, 9, None]
}

df_missing = pd.DataFrame(data_with_missing)
print("\n--- Yo'qotilgan ma'lumotli DataFrame ---")
print(df_missing)

# Qaysi ustunda nechta NaN borligini tekshirish
print("\n--- isnull().sum() ---")
print(df_missing.isnull().sum())

# NaN qiymatlari bor qatorlarni o'chirish
print("\n--- dropna() ---")
print(df_missing.dropna())

# Faqat 'cars' ustunida NaN bo'lgan qatorlarni o'chirish
print("\n--- dropna(subset=['cars']) ---")
print(df_missing.dropna(subset=['cars']))

# NaN qiymatlarni boshqa qiymat bilan to'ldirish
print("\n--- fillna() ---")
print(df_missing.fillna({"cars": "unknown", "counts": 0}))

# ------------------------------------------
# 6. Tashqi fayllar bilan ishlash (Titanic dataset, CSV, Excel, JSON)
# ------------------------------------------
# Eslatma: Titanic-Dataset.csv faylini Colab'dagi chap menyudagi "Files" bo'limiga yuklagan bo'lishingiz kerak
try:
    df_titanic = pd.read_csv("Titanic-Dataset.csv")
    print("\n--- Titanic Dataset Head ---")
    print(df_titanic.head())
except Exception as e:
    print("\nTitanic-Dataset.csv fayli topilmadi. Faylni yuklashni unutmang.")

# Excel faylga yozish
df.to_excel("test.xlsx")

# ------------------------------------------
# 7. Series va ro'yxatlar bilan ishlash
# ------------------------------------------
cars_series_direct = pd.Series(["BMW", "Tesla", "Ferrari"])
print("\n--- Pandas Series ---")
print(cars_series_direct)

cars_series = data_with_missing['cars']
print("\n--- Dictionary'dan 'cars' ro'yxatini olish ---")
print(cars_series)
```
