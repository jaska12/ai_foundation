# Pandas & NumPy: Student Performance Analysis

## 0. Kutubxonalarni import qilish

```python
import pandas as pd
import numpy as np
```

## 1. Datasetni yuklash

```python
df = pd.read_csv("Student_Performance.csv")

print("Dataset shape (rows, columns):", df.shape)
print("Column names:", df.columns.tolist())
```

```python
df.head()
```

## 2. Statistik ma'lumotlar va umumiy ma'lumot

```python
df.describe()
```

```python
df.info()
```

## 3. Fanlar bo'yicha eng yuqori va eng past baholar

```python
score_columns = ["math_score", "science_score", "english_score", "overall_score"]

print("Maximum scores by subject:")
print(df[score_columns].max())
```

```python
print("Minimum scores by subject:")
print(df[score_columns].min())
```

## 4. Korrelatsiya

```python
df[score_columns].corr()
```

## 5. Imtihondan o'tdi / o'tmadi (passed ustuni)

```python
# overall_score >= 50 bo'lsa "Yes", aks holda "No"
df["passed"] = np.where(df["overall_score"] >= 50, "Yes", "No")

df[["overall_score", "passed"]].head()
```

## 6. O'tganlar foizi (Pass rate)

```python
pass_rate = df["passed"].value_counts(normalize=True) * 100
print("Pass rate (%):")
print(pass_rate)
```

## 7. NumPy: o'rtacha, standart og'ish va Z-score

```python
scores = df[score_columns].values

print("Numpy Mean:", np.mean(scores, axis=0))
print("Numpy Std:", np.std(scores, axis=0))
```

```python
z_score = (scores - np.mean(scores, axis=0)) / np.std(scores, axis=0)

print("Z-scores (normalized numbers):")
print(z_score[:5])
```

## 8. Shartli qidiruv va filtrlash

```python
top_scorers = df[df["overall_score"] >= 90]
print("Top scored students (overall_score >= 90):", len(top_scorers))
```

```python
failed = df[df["passed"] == "No"]
print("Failed from exam:", len(failed))
```
