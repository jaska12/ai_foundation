```python
# ==========================================================
# PANDAS & NUMPY: STUDENT PERFORMANCE ANALYSIS (TO'LIQ TO'G'RILANGAN KOD)
# ==========================================================

import pandas as pd
import numpy as np

# 1. Datasetni yuklash (fayl nomi: Student_Performance.csv)
df = pd.read_csv("Student_Performance.csv")

print("Dataset shape (rows, columns):", df.shape)
print("Column names:", df.columns.tolist())
print("\nFirst 5 rows of the dataset:")
print(df.head())

# 2. Statistik ma'lumotlar va umumiy ma'lumot (info)
print("\nSummary statistics:")
print(df.describe())

print("\nData types and null info:")
print(df.info())

# 3. Fanlar bo'yicha eng yuqori va eng past baholar (fayldagi haqiqiy ustunlar)
score_columns = ["math_score", "science_score", "english_score", "overall_score"]

print("\nMaximum scores by subject:")
print(df[score_columns].max())

print("\nMinimum scores by subject:")
print(df[score_columns].min())

# 4. Korrelatsiya (Correlation)
print("\nCorrelation:")
print(df[score_columns].corr())

# 5. O'rtacha umumiy ballni hisoblash va shartli tekshiruv (Passed)
# overall_score asosida o'rtacha 50 dan baland bo'lsa "Yes", aks holda "No"
df["passed"] = np.where(df["overall_score"] >= 50, "Yes", "No")

print("\nNew columns preview (overall_score, passed):")
print(df[["overall_score", "passed"]].head())

# 6. Imtihondan o'tganlar foizi (Pass rate)
pass_rate = df["passed"].value_counts(normalize=True) * 100
print("\nPass rate:\n", pass_rate)

# 7. NumPy massiviga o'tkazish va Z-score hisoblash
scores = df[score_columns].values

print("\nNumpy Mean:", np.mean(scores, axis=0))
print("Numpy Std:", np.std(scores, axis=0))

# Z-score (normalize)
z_score = (scores - np.mean(scores, axis=0)) / np.std(scores, axis=0)
print("\nZ-scores (normalized numbers):")
print(z_score[:5])

# 8. Shartli qidiruv va filtrlash
top_scorers = df[df["overall_score"] >= 90]
print("\nTop scored students (overall_score >= 90):", len(top_scorers))

failed = df[df["passed"] == "No"]
print("Failed from exam:", len(failed))
```
